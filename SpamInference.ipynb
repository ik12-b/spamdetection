```python
import joblib
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')
model = joblib.load('/content/drive/MyDrive/model/spamprediction.pkl')
vectorizer = joblib.load('/content/drive/MyDrive/pre-processing/vectorizer.pkl')
def predict(text):
  text_vectorized = vectorizer.transform([text])
  prediction = model.predict(text_vectorized)
  id2label = {0: 'original', 1: 'spam'}
  prediction = [id2label[i] for i in prediction]
  return prediction
predict('hello')
```
