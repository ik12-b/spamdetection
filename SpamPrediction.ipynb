```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import numpy as np
import nltk
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from xgboost import XGBClassifier
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder
import joblib
import os
import time

df = pd.read_csv('/content/drive/MyDrive/Data/spam.csv', encoding='latin1')
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))


df['v2'] = df['v2'].astype(str).apply(lambda x: ' '.join([word for word in x.split() if word not in stop_words]))

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['v2'])

y = df['v1']


le = LabelEncoder()
y_encoded = le.fit_transform(y)


X_dense = X.toarray()

x_tr,x_ts,y_tr,y_ts = train_test_split(X_dense,y_encoded.astype(int), test_size=0.2, random_state=42)

model = XGBClassifier (
n_estimators=200,
learning_rate=0.05,
max_depth=6,
random_state=42
)

model = model.fit(x_tr,y_tr)

y_pred = model.predict(x_ts)


accuracy = accuracy_score(y_ts, y_pred)
print(f"Accuracy: {accuracy:.4f}")
print("Classification Report:")
print(classification_report(y_ts, y_pred))

joblib.dump(model, 'xgbC.pkl')
joblib.dump(vectorizer, 'vectorizer.pkl')

```
